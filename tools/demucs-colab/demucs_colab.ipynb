# ញែកភ្លេងដោយ Demucs លើ Google Colab (GPU) → គេហទំព័រ KhmerDub AI

Notebook នេះបើក **Demucs API** នៅលើ Colab (មាន GPU ឥតគិតថ្លៃ) រួចបើក tunnel ឲ្យគេហទំព័រ
ភ្ជាប់បាន។ លឿនជាងទូរស័ព្ទច្រើនដង ព្រោះ GPU ញែកសំឡេងបានក្នុងរយៈពេលតិចជាង ១ ភាគ ១០
នៃប្រវែងសំឡេង។

**របៀបប្រើ (រត់តាមលំដាប់ បន្ទាត់ទី ១ ដល់ ៣)៖**

1. `Runtime → Change runtime type → T4 GPU` (សំខាន់ — បើគ្មាន GPU វាយឺតដូច CPU)
2. រត់ cell ទី ២ (ពិនិត្យ GPU)
3. រត់ cell ទី ៣ (បើក API + tunnel) — វាបង្ហាញ **URL** និង **API Key**
4. ចម្លង URL + API Key ទៅកាត «ញែកភ្លេង · Demucs API» ក្នុងផ្ទាំងស្ទូឌីយោ រួចចុច «រក្សាទុក និងសាកល្បង»

⚠️ ទុក tab នេះចោលបើកចុះ។ បិទ tab ឬទុកចោល (idle) ប្រហែល ៩០ នាទី = session ដាច់ ហើយ
URL នោះស្លាប់ — បើក cell ទី ៣ ម្ដងទៀត រួច paste URL ថ្មីក្នុងកាត។


```python
# ជំហានទី ១ — ពិនិត្យថា GPU ត្រូវបានបើក
# បើឃើញ "គ្មាន GPU" សូមធ្វើ៖ Runtime → Change runtime type → T4 GPU
!nvidia-smi -L || echo "គ្មាន GPU — សូមបើក T4 GPU ក្នុង Runtime → Change runtime type"
import torch
print("GPU ប្រើបាន:", torch.cuda.is_available())
```

```python
# ជំហានទី ២ — បើក Demucs API + tunnel (បញ្ជាតែមួយ)
#
# API Key៖ ដាក់ key ផ្ទាល់ខ្លួននៅបន្ទាត់ខាងក្រោម (កែអក្សរនៅក្នុងសញ្ញាក្បៀស) ដើម្បីកុំឲ្យ
# វាប្តូររាល់ពេលបើក session ថ្មី — រួចអ្នកគ្រាន់តែប្តូរ **URL** ក្នុងកាតលើគេហទំព័រ។
import os
os.environ["DEMUCS_API_KEY"] = "khmerdub-colab-key"   # ➜ ប្តូរជាអក្សរសម្ងាត់ផ្ទាល់ខ្លួនរបស់អ្នក

!curl -fsSL https://raw.githubusercontent.com/kitreaksa78-debug/Sal/main/tools/demucs-colab/colab-start.sh | sh
```

```python
# ជំហានទី ៣ — មើល URL + Key ម្តងទៀតពេលណាក៏បាន (API និង tunnel នៅដើរ background)
import json, os, urllib.request
print("URL  :", open("/content/demucs-api/tunnel-url.txt").read().strip() if os.path.exists("/content/demucs-api/tunnel-url.txt") else "(មិនទាន់មាន)")
print("Key  :", os.environ.get("DEMUCS_API_KEY"))
port = '8001'
if os.path.exists('/content/demucs-api/port.txt'):
    port = open('/content/demucs-api/port.txt').read().strip() or port
print("ស្ថានភាព API:", json.load(urllib.request.urlopen(f"http://127.0.0.1:{port}/")))
```
